# Download Indian Equity Price Data

This notebook downloads daily price data for Nifty and a starter universe of large Indian stocks.

The goal is to create the raw price dataset needed for the Indian Market Internals project.

```python
!pip -q install yfinance pandas numpy matplotlib
```

```python
import pandas as pd
import numpy as np
import yfinance as yf
from pathlib import Path

pd.set_option("display.max_columns", 50)
```

```python
START_DATE = "2018-01-01"

INDEX_TICKER = "^NSEI"  # Nifty 50 index on Yahoo Finance

STOCK_TICKERS = [
    "RELIANCE.NS",
    "TCS.NS",
    "HDFCBANK.NS",
    "ICICIBANK.NS",
    "INFY.NS",
    "HINDUNILVR.NS",
    "ITC.NS",
    "LT.NS",
    "SBIN.NS",
    "BHARTIARTL.NS",
    "AXISBANK.NS",
    "KOTAKBANK.NS",
    "BAJFINANCE.NS",
    "ASIANPAINT.NS",
    "MARUTI.NS",
    "HCLTECH.NS",
    "SUNPHARMA.NS",
    "TMCV.NS",
    "TITAN.NS",
    "ULTRACEMCO.NS",
    "NESTLEIND.NS",
    "POWERGRID.NS",
    "NTPC.NS",
    "ONGC.NS",
    "COALINDIA.NS",
]

TICKERS = [INDEX_TICKER] + STOCK_TICKERS

len(TICKERS), TICKERS[:5]
```

```python
raw = yf.download(
    tickers=TICKERS,
    start=START_DATE,
    auto_adjust=True,
    group_by="ticker",
    multi_level_index=True,
    progress=False,
    threads=True
)

raw.head()
```

```python
downloaded_tickers = sorted(set(raw.columns.get_level_values(0)))

missing_tickers = [ticker for ticker in TICKERS if ticker not in downloaded_tickers]

print("Downloaded tickers:", len(downloaded_tickers))
print("Missing tickers:", missing_tickers)
```

```python
close_prices = pd.DataFrame()

for ticker in TICKERS:
    if ticker in downloaded_tickers:
        close_prices[ticker] = raw[ticker]["Close"]

close_prices.index = pd.to_datetime(close_prices.index)
close_prices.index.name = "Date"

close_prices = close_prices.rename(columns={INDEX_TICKER: "NIFTY"})
close_prices = close_prices.dropna(how="all")

close_prices.tail()
```

```python
summary = pd.DataFrame({
    "first_valid_date": close_prices.apply(lambda s: s.first_valid_index()),
    "last_valid_date": close_prices.apply(lambda s: s.last_valid_index()),
    "missing_pct": close_prices.isna().mean().round(4)
})

summary.sort_values("missing_pct", ascending=False).head(10)
```

```python
output_dir = Path("/content/indian-market-internals-output")
output_dir.mkdir(parents=True, exist_ok=True)

close_prices.to_csv(output_dir / "close_prices_starter_universe.csv")

print("Saved file:")
print(output_dir / "close_prices_starter_universe.csv")
```
